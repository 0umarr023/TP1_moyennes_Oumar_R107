CALCUL DES MOYENNES UE 


```python
#Algorithme 


# On demande le nombre de modules
#Le int sert a transfomer le input en nombres on met alors inputpour afficher la chaines de caractere nombre de modules dans la variables nb_modules

nb_moyennes = int(input("Nombre de modules : "))
                 
#On programme mtn les totaux qui sont base sur la moyennes pondérés enfaite d'abord pour chaque total celui de ue1,..,ue3 on calcul la note mutilplie par le coefficient qui apres vont etre divise par la somme des coefficient de chaque ue est ca donne la moyenne de l'ue1 par exemple ils doivent commencé a 0 

total1 = 0
total2 = 0
total3 = 0
coef1 =  0
coef2 =  0
coef3 =  0

#Pour chaque modules on vas demander la note et ces 3 coefficients 

for i in range(nb_moyennes):
    note = float(input("rentrez la note: "))
    c1 =  float(input("rentrez le coefficient ue1: "))
    c2 =  float(input("rentrez le coefficient ue2: "))
    c3 =  float(input("rentrez le coefficient ue3: "))

    total1 = total1 + note * c1
    total2 = total2 + note * c2
    total3 = total3 + note * c3
    coef1 = coef1 + c1
    coef2 = coef2 + c2
    coef3 = coef3 + c3

ue1 = total1/coef1
ue2 = total2/coef2
ue3 = total3/coef3

print ("UE1",ue1)
print ("UE2",ue2)
print ("UE3",ue3)

# Cette partie quand a elle sert a dire que on part du proincipes que e nombre ue valide est egal a 0 
#si ue1 est egal a 0 on ajoute un a nombre UE valide ce qui devient 1 et ainsi on a valide une UE est ainsi de suite 
nb_ue_validees = 0
if ue1 >= 10:
    nb_ue_validees = nb_ue_validees + 1
if ue2 >= 10:
    nb_ue_validees = nb_ue_validees + 1
if ue3 >= 10:
    nb_ue_validees = nb_ue_validees + 1

# Ici on montre que si sompbre ue valide egal 3 c'est bon et on affiche année validée passage en but2 , par a suite 
#si nombre ue validées egal 2 ET LE MINIMUM DES ue& JUSQU'a 3 superieur ou egal a 8 ainsi de suite pour absolument chaque situation un message sera envenyé apres le calcul de la moyenne 
if nb_ue_validees == 3:
    print("Année validée, passage en BUT2")
elif nb_ue_validees == 2 and min(ue1, ue2, ue3) >= 8:
    print("Année non validée, mais passage en BUT2")
elif nb_ue_validees == 2:
    print("Année non validée, redoublement possible")
else:
    print("Année non validée, pas de passage en BUT2")
    


                
```
