```python
#Algorithme 


# On demande le nombre de modules
#Le int sert a transfomer le input en nombres on met alors inputpour afficher la chaines de caractere nombre de modules dans la variables nb_modules

nb_moyennes = int(input("Nombre de modules : "))
                 
#On programme mtn les totaux qui sont base sur la moyennes pondérés enfaite d'abord pour chaque total celui de ue1,..,ue3 on calcul la note mutilplie par le coefficient qui apres vont etre divise par la somme des coefficient de chaque ue est ca donne la moyenne de l'ue1 par exemple ils doivent commencé a 0 

total1 = 0
total2 = 0
total3 = 0
coef1 =  0
coef2 =  0
coef3 =  0

#Pour chaque modules on vas demander la note et ces 3 coefficients 

for i in range(nb_moyennes):
    note = float(input("rentrez la note: "))
    c1 =  float(input("rentrez le coefficient ue1: "))
    c2 =  float(input("rentrez le coefficient ue2: "))
    c3 =  float(input("rentrez le coefficient ue3: "))

    total1 = total1 + note * c1
    total2 = total2 + note * c2
    total3 = total3 + note * c3
    coef1 = coef1 + c1
    coef2 = coef2 + c2
    coef3 = coef3 + c3

ue1 = total1/coef1
ue2 = total2/coef2
ue3 = total3/coef3

print ("UE1",ue1)
print ("UE2",ue2)
print ("UE3",ue3)

# Cette partie quand a elle sert a dire que on part du proincipes que e nombre ue valide est egal a 0 
#si ue1 est egal a 0 on ajoute un a nombre UE valide ce qui devient 1 et ainsi on a valide une UE est ainsi de suite 
nb_ue_validees = 0
if ue1 >= 10:
    nb_ue_validees = nb_ue_validees + 1
if ue2 >= 10:
    nb_ue_validees = nb_ue_validees + 1
if ue3 >= 10:
    nb_ue_validees = nb_ue_validees + 1

# Ici on montre que si sompbre ue valide egal 3 c'est bon et on affiche année validée passage en but2 , par a suite 
#si nombre ue validées egal 2 ET LE MINIMUM DES ue& JUSQU'a 3 superieur ou egal a 8 ainsi de suite pour absolument chaque situation un message sera envenyé apres le calcul de la moyenne 
if nb_ue_validees == 3:
    print("Année validée, passage en BUT2")
elif nb_ue_validees == 2 and min(ue1, ue2, ue3) >= 8:
    print("Année non validée, mais passage en BUT2")
elif nb_ue_validees == 2:
    print("Année non validée, redoublement possible")
else:
    print("Année non validée, pas de passage en BUT2")
    


                
```

```python
# VERSION 2 - Avec listes et compréhension de liste

# Création de la liste des matières avec compréhension de liste
matieres = [f"R{i}" for i in range(101,116)] + [f"SAE{i}" for i in range(11,17)]
print("Matières :", matieres)
print(f"Total : {len(matieres)} matières\n")

# Initialisation des listes pour les totaux et coefficients
totaux = [0, 0, 0]  # Pour UE1, UE2, UE3
coefficients = [0, 0, 0]

# Pour chaque matière
for matiere in matieres:
    print(f"\n--- {matiere} ---")
    note = float(input(f"Note pour {matiere} : "))

    # Pour chaque UE
    for ue in range(3):
        c = float(input(f"Coefficient UE{ue+1} : "))
        totaux[ue] = totaux[ue] + note * c
        coefficients[ue] = coefficients[ue] + c

# Calcul des moyennes avec compréhension de liste
moyennes = [totaux[i] / coefficients[i] for i in range(3)]

print("\n" + "="*40)
print(f"UE1 : {moyennes[0]:.2f}")
print(f"UE2 : {moyennes[1]:.2f}")
print(f"UE3 : {moyennes[2]:.2f}")
print("="*40)

# Compte les UE validées avec compréhension de liste
nb_ue_validees = sum(1 for m in moyennes if m >= 10)

# Logique du jury
if nb_ue_validees == 3:
    print("✓ Année validée, passage en BUT2")
elif nb_ue_validees == 2 and min(moyennes) >= 8:
    print("⚠ Année non validée, mais passage en BUT2")
elif nb_ue_validees == 2:
    print("✗ Année non validée, redoublement possible")
else:
    print("✗ Année non validée, pas de passage en BUT2")
```

```python
# VERSION 4 - Graphique avec gestion d'erreur

import matplotlib.pyplot as plt

# Liste des matières
matieres = [f"R{i}" for i in range(101,116)] + [f"SAE{i}" for i in range(11,17)]
print("Matières :", matieres)
print()

# Initialisation
totaux = [0, 0, 0]
coefficients = [0, 0, 0]

# Pour chaque matière
for matiere in matieres:
    # Demander la note avec vérification
    while True:
        try:
            note = float(input(f"Note {matiere} : "))
            break
        except ValueError:
            print(" Erreur ! Rentre un CHIFFRE ")

    # Pour chaque UE
    for ue in range(3):
        while True:
            try:
                c = float(input(f"  Coefficient UE{ue+1} : "))
                break
            except ValueError:
                print(" Erreur ! Rentre un CHIFFRE")

        totaux[ue] = totaux[ue] + note * c
        coefficients[ue] = coefficients[ue] + c

# Calcul des moyennes
moyennes = [totaux[i] / coefficients[i] for i in range(3)]

print("\n" + "="*40)
print(f"UE1 : {moyennes[0]:.2f}")
print(f"UE2 : {moyennes[1]:.2f}")
print(f"UE3 : {moyennes[2]:.2f}")
print("="*40 + "\n")

# GRAPHIQUE avec matplotlib
noms_ue = ["UE1", "UE2", "UE3"]
couleurs = ['green' if m >= 10 else 'red' for m in moyennes]

plt.figure(figsize=(8, 5))
plt.bar(noms_ue, moyennes, color=couleurs, alpha=0.7)
plt.axhline(y=10, color='red', linestyle='--', linewidth=2, label='Validation (10)')
plt.ylabel("Moyenne", fontsize=12)
plt.title("Moyennes par UE", fontsize=14, fontweight='bold')
plt.ylim(0, 20)
plt.legend(fontsize=10)
plt.grid(axis='y', alpha=0.3)
plt.show()

# Décision
validees = sum(1 for m in moyennes if m >= 10)
print("\n" + "="*40)
if validees == 3:
    print(" Année validée, passage en BUT2")
elif validees == 2 and min(moyennes) >= 8:
    print(" Année non validée, mais passage en BUT2")
elif validees == 2:
    print(" Année non validée, redoublement possible")
else:
    print(" Année non validée, pas de passage en BUT2")
print("="*40)
```
